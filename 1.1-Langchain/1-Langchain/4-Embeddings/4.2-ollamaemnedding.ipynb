#### Ollama
Ollama supports embedding models, making it possible to build retrieval augmented generation (RAG) applications that combine text prompts with existing documents or other data.

```python
## OllamaEmbeddings is a class that provides an interface to generate embeddings using the Ollama model. 
## It can be used to create vector representations of text data, which can then be used for various applications such as similarity search, clustering, etc.
from langchain_ollama import OllamaEmbeddings
```

```python
## By default, it uses the llama2 model, but you can specify a different model by passing the model name as an argument to the OllamaEmbeddings class.
embeddings=(
    OllamaEmbeddings(model="gemma:2b")
)

embeddings
```

```python
## embeddings.embed_documents is a method that takes a list of documents as input and returns their corresponding embeddings.
r1 = embeddings.embed_documents(
    [
       "Alpha is the first letter of Greek alphabet",
       "Beta is the second letter of Greek alphabet", 
    ]
)

print(len(r1[0]))
r1[1]
```

```python
## To check the answer of the query, we can use the embed_query method which takes a query as input and returns its corresponding embedding.
embeddings.embed_query("What is the second letter of Greek alphabet ")
```

```python
## Other Embedding Models
## https://ollama.com/blog/embedding-models
embeddings = OllamaEmbeddings(model = "mxbai-embed-large")
text = "This is a test document."

## Generate the embedding for the query
query_result = embeddings.embed_query(text)
query_result
```
